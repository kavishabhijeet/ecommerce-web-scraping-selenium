```python
page_count = 1

while True:

    print("Scraping page:", page_count)

    next_buttons = driver.find_elements(
        "css selector",
        "li.next a"
    )

    if not next_buttons:
        break

    next_buttons[0].click()
    page_count += 1

print("Total pages:", page_count)
```

```python
driver.get("https://books.toscrape.com/")
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

products = []

while True:

    books = driver.find_elements(
        By.CSS_SELECTOR,
        "article.product_pod"
    )

    print("Books on current page:", len(books))

    for book in books:

        title = book.find_element(
            By.CSS_SELECTOR, "h3 a"
        ).get_attribute("title")

        price = book.find_element(
            By.CSS_SELECTOR, ".price_color"
        ).text

        rating = book.find_element(
            By.CSS_SELECTOR, ".star-rating"
        ).get_attribute("class")

        availability = book.find_element(
            By.CSS_SELECTOR, ".instock.availability"
        ).text.strip()

        product = {
            "Title": title,
            "Price": price,
            "Rating": rating,
            "Availability": availability
        }

        products.append(product)

    next_buttons = driver.find_elements(
        By.CSS_SELECTOR,
        "li.next a"
    )

    if not next_buttons:
        break

    next_buttons[0].click()

    WebDriverWait(driver, 5).until(
        EC.presence_of_element_located(
            (By.CSS_SELECTOR, "article.product_pod")
        )
    )

print("Total products:", len(products))
```

## 📊 Price Analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Price",
    bins=30,
    kde=True
)

plt.title("Distribution of Book Prices", fontsize=16, fontweight="bold")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")

plt.tight_layout()
plt.show()
```

### Observation:
Book prices range from £10.00 to £59.99, with an average price of approximately £35.07. The distribution is fairly spread across the price range, with books available at both lower and higher price points rather than being concentrated around a single price range.

### Business Insight

Business Insight:
The wide price range indicates that the marketplace offers books across different price segments. This can support a diverse pricing strategy, catering to both budget-conscious and premium-segment customers.

## Rating Distribution ⭐

```python
df["Rating"].value_counts().sort_index()
```

```python
plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=df,
    x="Rating"
)

plt.title("Distribution of Book Ratings", fontsize=16, fontweight="bold")
plt.xlabel("Rating")
plt.ylabel("Number of Books")

for container in ax.containers:
    ax.bar_label(container, fontsize=10)

plt.tight_layout()
plt.show()
```

### Observation:
The distribution of book ratings is relatively balanced across all five rating levels. 1-star books are the highest at 226, while 4-star books are the lowest at 179. Ratings of 2 and 5 stars each have 196 books, while 3-star books account for 203 books.

### Business Insight

Business Insight:
The balanced rating distribution indicates that the catalog contains books with varying levels of customer appeal, rather than being dominated by highly rated books. This can help a marketplace maintain a diverse catalog while using higher-rated books for recommendations and promotional campaigns.

## Top 10 Most Expensive Books 💷

```python
top_10_expensive = df.nlargest(10, "Price")[["Title", "Price", "Rating"]]

top_10_expensive
```

```python
plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=top_10_expensive,
    x="Price",
    y="Title",
    hue="Title",
    palette="viridis",
    legend=False
)

plt.title(
    "Top 10 Most Expensive Books",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Price (£)")
plt.ylabel("Book Title")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="£%.2f",
        padding=3,
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

### Observation:
The top 10 most expensive books are priced within a very narrow range of £59.45 to £59.99. The Perfect Play (Play by Play #1) is the most expensive at £59.99, while The Improbability of Love and The Man Who Mistook His Wife for a Hat and Other Clinical Tales are priced at £59.45.

### Business Insight:
The concentration of prices around £60 indicates that the highest-priced books follow a premium pricing range rather than having large price differences. These books could be highlighted as premium offerings, while their ratings and availability can be considered alongside price when evaluating their attractiveness to customers.
